```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats
```

С наступающим новым годом!

```python
df = pd.read_csv('avia_price_data.csv')
df.head(3)
```

```python
n = len(df)
X = df['avia_payments_rub']
Y = df['avia_payments_cnt']

R = X.mean() / Y.mean()

sigmaW2 = (X.var(ddof=1) - 2*R*np.cov(X, Y, ddof=1)[0,1] + R**2 * Y.var(ddof=1))
sigmaR2 = sigmaW2 / (n * Y.mean()**2)
print("дисперсия R по всей выборке: ", sigmaR2)
print("дисперсия W (линеаризация): ", sigmaR2 * n)
```

Теперь проверим с помощью линеаризации

```python
W = R + (1 / Y.mean()) * (X - R * Y)
var_W = W.var(ddof=1)
var_W
```
