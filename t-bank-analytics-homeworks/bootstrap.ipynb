```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import ttest_ind
from tqdm import tqdm
```

```python
df = pd.read_csv("data_with_test_results.csv")

grouped_df = df.groupby(["user_id", "group"]).agg(
    orders_cnt=("orders_cnt", "sum")
).reset_index()

dfA = grouped_df[grouped_df["group"] == "A"]
dfB = grouped_df[grouped_df["group"] == "B"]
```

```python
t_value, p_value = ttest_ind(dfA["orders_cnt"], dfB["orders_cnt"])
print("T-value:", t_value)
print("P-value:", p_value)
```

```python
def bootstrap_means(data_A, data_B, n=10000):
    differences = []
    for _ in range(n):
        bootstrap_A = np.random.choice(data_A, size=len(data_A), replace=True)
        bootstrap_B = np.random.choice(data_B, size=len(data_B), replace=True)
        
        diff = np.mean(bootstrap_B) - np.mean(bootstrap_A)
        differences.append(diff)
    return np.array(differences)

bootstrap_diffs = bootstrap_means(dfA["orders_cnt"], dfB["orders_cnt"])
lower = np.percentile(bootstrap_diffs, 2.5)
upper = np.percentile(bootstrap_diffs, 97.5)
p_value_bootstrap = np.mean(bootstrap_diffs <= 0)
```

```python
print(f"Бутстрап: 95% доверительный интервал для разницы средних = [{lower:.4f}, {upper:.4f}]")
print(f"Бутстрап p-value = {p_value_bootstrap:.4f}")
```
