```python
import numpy as np
import scipy.stats as stats
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import t
```

```python
df = pd.read_csv("sex_bmi_smokers.csv")
df.head()
```

```python
smokers = df[df['smoker'] == "yes"]
healthy = df[df['smoker'] == "no"]

s_mean = smokers['bmi'].mean()
h_mean = healthy['bmi'].mean()
all_mean = df['bmi'].mean()

s = smokers['bmi'].tolist()
h = healthy['bmi'].tolist()
```

```python
var_s = len(s) * (s_mean - all_mean)**2
var_h = len(h) * (h_mean - all_mean)**2
ssb = var_s + var_h

var_s_in = sum((x - s_mean)**2 for x in s)
var_h_in = sum((x - h_mean)**2 for x in h)
ssw = var_s_in + var_h_in

print('ssb: ', ssb)
print('ssw: ', ssw)
```

```python
k = 2
sz = len(df)
msb = ssb / (k - 1)
msw = ssw / (sz - k)
F = msb / msw
print('F: ', F)
```

```python
p_value = 1 - stats.f.cdf(F, dfn=k-1, dfd=sz-k)
```

```python
print(f"Среднее BMI курящих: {s_mean:.4f}")
print(f"Среднее BMI некурящих: {h_mean:.4f}")
print(f"Общее среднее: {all_mean:.4f}")
print(f"Размеры групп: курящие={len(s)}, некурящие={len(h)}")
print(f"F-статистика: {F:.4f}, p-value: {p_value:.4f}")
```

Не отвергаем H0. Средние BMI не различаются.
